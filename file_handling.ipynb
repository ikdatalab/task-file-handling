__Question 2:__ Transaction Revenue Calculator (Read & Parse)
Using the `daily_sales.txt` file created in Question 1:

1. Open and read the file line by line using Python (skip the header).
2. Extract the item name, quantity, and price.
3. Calculate the total revenue in PKR ($\text{Quantity} \times \text{Price\_PKR}$).
4. Print a summary in this exact format:

__Plaintext__

`Total Items Sold: X`

`Total Revenue Generated: PKR Y`

```python
total_revenue = 0
total_items = 0
with open('daily_sales.txt', 'r') as a:
    next(a)
    for line in a:
        data = line.strip().split(',')
        data[0]
        data[1]
        data[2]
        revenue = int(data[1]) * int(data[2])
        total_revenue += revenue
        total_items += int(data[1])

print(f"Total Items Sold: {total_items}")
print(f"Total Revenue Generated: PKR {total_revenue}")
```

__Question 3:__ Word Search & Frequency Counter (Search Algorithm)
Create a file named `log.txt` containing the following text:

_"system error on server 1. database connection failed. error details: timeout error."_

Write a Python program that:
1. Reads `log.txt`.
2.  Searches for the keyword `"error"` (case-insensitive).
3. Counts how many times `"error"` appears in the text.
4. If found, prints `Found 'error' X times in log.txt.` If not found, prints `Keyword not found.`

```python
with open ('log.txt','w') as a:
    a.write ("system error on server 1. database connection failed. error details: timeout error.")

keyword = 0
with open ('log.txt','r') as b:
    for line in b:
       keyword += line.lower().count('error')

if keyword> 0:
    print(f"Found 'error' {keyword} times in log.txt.")
else:
    print("Keyword not found.") 
```

__Question 4:__ Comma-Separated Number Analysis (Parsing & Logic)
Create a file named `numbers.txt` with the following content on a single line:
`120, 45, 89, 230, 67, 102, 34, 500, 11`

Write a Python script that:

1. Reads the contents of `numbers.txt`.
2. Parses the string into a list of integers using `.split(',')`.
3. Counts how many numbers are greater than 100.
4. Writes the results into a new file named `filtered_summary.txt` in the following format:

`Original Count: 9`

`Numbers > 100: 4`

`Max Number: 500`

```python
number = (120, 45, 89, 130, 67, 102, 34, 500, 11)
with open('numbers.txt','w') as a:
    a.write(",".join(map(str, number)))

number_list = []
with open('numbers.txt','r') as b:
    data = b.read()
    for item in data.split(","):
        number_list.append(int(item.strip()))

original_count = len(number_list)
number_gre = len([num for num in number_list if num > 100])
max_number = max (number_list)

with open ('filtered_summary','w') as c:
    c.write (f"Original Count: {original_count}\n")
    c.write(f"Numbers > 100: {number_gre}\n")
    c.write (f"Max Number: {max_number}")

print("Parsing & Analysis Complete!")
print(f"Parsed List: {number_list}")
print(f"Original Count: {original_count}")
print(f"Numbers > 100: {number_gre}")
print(f"Max Number: {max_number}")
```

__Question 5:__ File Cleanup & Safety (__`OS` Module__)

Write a Python script that checks if a file named `temp_cache.txt` exists:

1. f it exists, remove it using the os module and print `"temp_cache.txt has been deleted successfully."`

2. If it does not exist, create `temp_cache.txt`, write `"temporary data"` into it, and print `"temp_cache.txt was missing, so it was created."`



```python
import os
file_name = "temp_cache.txt"

if os.path.exists(file_name):
    os.remove(file_name)
    print("temp_cache.txt has been deleted successfully.")

else:
    with open ('temp_cache.txt','w') as a:
        a.write("temporary data")
    print("temp_cache.txt was missing, so it was created")
```

__Question 6:__ Real-World Scenario — Customer Data Anonymizer

A local fintech firm receives customer records in `customers.txt`:

__DATA__

`Name,CNIC,City`

`Ali Ahmed,42101-1234567-1,Karachi`

`Sara Khan,35202-7654321-2,Lahore`

`Usman Raza,41303-9876543-3,Hyderabad`

```python
with open ('customers.txt','w') as a:
    a.write(f"""Name,CNIC,City
Ali Ahmed,42101-1234567-1,Karachi
Sara Khan,35202-7654321-2,Lahore
Usman Raza,41303-9876543-3,Hyderabad""")

with open ('customers.txt','r') as b, open ('anonymizer_customers.txt','w') as c:
    c.write("Name,CNIC,City\n")
    next(b)
    for line in b:
        data = line.strip().split(',')
        name = data[0]
        cnic = data[1]
        city = data[2]

        cnic_parts =  cnic.split('-')
        masked_cnic = f"*****-*******-{cnic_parts[2]}"

        c.write(f"{name},{masked_cnic},{city}\n")
```
